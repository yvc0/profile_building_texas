# Python Variables and Data Types

This notebook explains Python variables, data types, and includes problem statements with commented code examples.

## What is a Variable?

A variable is a name that stores a value. In Python, you can assign values to variables using the `=` operator. Variables can store different data types such as numbers, text, and boolean values.

```python
# Assign values to variables
age = 20
name = 'Deep'
is_student = True
print('Name:', name)
print('Age:', age)
print('Is student:', is_student)
```

## Data Types in Python

Common Python data types include: `int`, `float`, `str`, `bool`, `list`, `tuple`, `dict`, and `set`. Variables can change type because Python is dynamically typed.

```python
# Display data types of variables
x = 45
y = 3.14
z = 'Python'
print(type(x))
print(type(y))
print(type(z))
```

## Problem Statement 1: Create a student profile

Define variables for a student's name, age, course, and grade. Then print a short profile message using those variables.

```python
# Problem 1 solution
student_name = 'Sara'
student_age = 18
course = 'Python Basics'
grade = 'A+'
print('Student:', student_name)
print('Age:', student_age)
print('Course:', course)
print('Grade:', grade)
```

## Type Conversion

Sometimes you need to convert values between types. Python provides functions like `int()`, `float()`, and `str()` for conversion.

```python
# Convert between different data types
value = '2026'
year = int(value)
print('String value:', value)
print('Converted to integer:', year)
print('Year plus 1 =', year + 1)
```

## Problem Statement 2: Calculate total cost

Write a script that defines item price and quantity as variables, converts them to the correct type when needed, and prints the total cost.

```python
# Problem 2 solution
price = 29.99
quantity = 3
total_cost = price * quantity
print('Price per item:', price)
print('Quantity:', quantity)
print('Total cost:', total_cost)
```

## Multiple Assignment and Swapping

Python allows assigning multiple variables in a single statement and swapping values without a temporary variable.

```python
# Multiple assignment example
a, b, c = 1, 2, 3
print('a, b, c =', a, b, c)

# Swap values
a, b = b, a
print('After swap a, b =', a, b)
```

## Constants

Constants are variables that should not change, typically written in uppercase by convention. Python does not enforce constants, but the naming helps communicate intent.

```python
# Constant example
PI = 3.14159
print('Value of PI =', PI)
radius = 5
area = PI * radius * radius
print('Area of circle:', area)
```
