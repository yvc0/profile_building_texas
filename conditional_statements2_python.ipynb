```python
# Atm tranasaction using dynamic values
print("welcome to abcd bank")
print("please insert ATM card")
pin_number = int(input("enter a pin:"))
amount = 50000
if (pin_number==7891):
    print("options \n1.check balence \n2.withdraw")
    choose = int(input("choose above options:"))
    if(choose==1):
        print("available balence is ",amount)
    elif(choose==2):
        with_draw=int(input("how much with draw amount do you want:"))
        if (with_draw<50000 and with_draw>=100):
            print("successfully with draw amount is ",with_draw)
            print("available balence is ",amount-with_draw)
        else:
            print("enter in suffucient funds or wrong amount you enter")
    else:
        print("please check correct option")
else:
    print("please enter correct pin number")
            
```
