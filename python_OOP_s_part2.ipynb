```python
class ABC: # this ABC is parent class
    name = 'venkatesh'
    address = 'hyderabad'

class ABCD(ABC):# here ABCD is child class getting or acquring properties from parent class
    email = 'venkatesh@gmail.com'
    contact = '546789'
class ABCDE(ABCD):
    age = 29
    salary = 45678
object1 = ABCDE()
print(object1.name)
print(object1.address)
print(object1.email)
print(object1.age)
```

```python
# best example for single inheritance
class CarBrands: # parent classs
    brand1 = 'BENZ'
    brand2 = 'AUDI'
    brand3 = 'BMW'
    brand4 = 'TATA'
    brand5 = 'SUZUKI'
class VehicleModel(CarBrands): # child
    model1 = 'E-Class'
    model2 = 'Q3'
    model3 = '3 series'
    model4 = 'Nexon'
    model5 = 'Swift'
    
cardata = VehicleModel()
print(cardata.brand1+' and top model is'+cardata.model1)
print(cardata.brand2+' and top model is'+cardata.model2)
print(cardata.brand3+' and top model is'+cardata.model3)
print(cardata.brand4+' and top model is'+cardata.model4)
print(cardata.brand5+' and top model is'+cardata.model5)
```
