#### Dictionaries
- dictionaries is collection of unorder , unindexed items enclosed by curly braces {item1,item2,..}
- unindex there is no index
- there is no order
- item is a key and value pair-{key:value}
- example1 = {'name':'venkatesh','age':28,'location':'hyderbad'}
- keys - ('name','age','location')
- values - ('venkatesh',28,'hyderabad')

```python
venkat_data = {'name':'venkatesh','age':28,'location':'hyderabd','contact':'+9112345678890'}
joseph_data = {'name':'joseph','age':25,'location':'mumbai','conatct':"not defined"}
print(type(venkat_data))
print(type(joseph_data))
```
