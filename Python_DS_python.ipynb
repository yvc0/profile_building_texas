#### python Data Structure
- data structures simple we can define as collection data
1. string

    - some examples
        1. name="python"
        2. dob = '1995'
        3. string1 = "python@1991"
    - multiline string
        - example1 = """line1

        6. digit

2. list
    - list is collection of elements or collection of items enclosed by squre brackets[]
    - example
        1. example1 = [1,2,3,4,5,6]
        2. example2 = ["venkat","joseph","aniket","indu","sonam"]
        3. example3 = [1,23,4,"python","java","c","c++"]
    - list is mutable
        - mutable means we can edit,update,delete,insert the items
    - methods
        1. append
        2. extend
        3. insert
        4. update
        5. pop
        6. remove 
3. tuple
    - tuple is collection of items enclosed by paranthesis()
    - examples
        1. example1 = (1,2,3,4,5,6)
        2. example2 = ("tanjeela","bhavya","divya")
        3. example3 = (1,2,12.45,12.67,"python","django","flask","data science")
    - tuple is immutable
        - once create a tuple after we can't do modifications
    - methods
        1. index
        2. count
4. dictionaries
    - dictionaries is collection of items(key and value pair is a item) enclosed by curly braces
    - examples
        1. example1 = {"name":"venkat","age":28,"location":"hyderabad"}
        2. example2 = {1:"venkatesh",2:"joseph",3:"aniket"}
    - dict is a mutable
    - methods
        1. update
        2. pop
        3. popitems
        4. append......
5. sets
    - collection of unorder items enclosed by curly braces is called sets
    - examples
        1. example1= {1,2,3,4,5}
        2. example2={1.2,1.3,1.4,1.5}
        3. example3= {"name1","name2"}
