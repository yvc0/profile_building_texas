```python
# mutable and immutable
# mutable - we can edit we can delete we can insert - list
# immutable - we can not edit we can not delete we can not insert - tuple
lang = ['python','javayyy','protyrging']
lang1 = ('python','javayyy','protyrging')
print(type(lang))
print(type(lang1))
```

```python
lang[1] = 'java'
lang[2] = 'programming'
print(lang)
```

```python
lang1[1] = 'java'
lang1[2] = 'programming'
print(lang1)
```

```python
# tuple concatenate - adding multiple tuples
names = ('venkat','joseph','aniket')
ages = (28,20,21)
location = ('hyderabad','mumbai','pune')
print(names + ages + location)
```

```python
# accessing elements from tuple
# slicing
names = ("venkat",'joseph',"aniket",'bharath','salman')
# accesing elements from tuple using []
# forward and backward index
print(names[2])
print(names[-5])
print(names[3:4])
```
