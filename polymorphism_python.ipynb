```python
class Cars:
    def wheels(self):
        print('car having 4 wheels')
    def fuel(self):
        print('car used by petrol')
class Bikes:
    def wheels(self):
        print('bike having 2 wheels')
    def fuel(self):
        print('bikes used by petrol')
car = Cars()
bike = Bikes()
def Vehicledata(data):
    data.wheels()
    data.fuel()
Vehicledata(car)
Vehicledata(bike)
```

```python
# polymorphism example2
# for loop,objects, functions

class Parent:
    def names(self):
        print('father name is abcde')
        print('mother name is yuiooo')
    def ages(self):
        print('father age is 50')
        print('mother age is 45')
    def contact(self):
        print('father contact number is 23456789')
        print('mother contact number is 98765430')
class Child:
    def names(self):
        print('daughter name is fghjk')
        print('son name is jhgfd')
    def ages(self):
        print('daughter age is 20')
        print('son age is 16')
    def contact(self):
        print('daughter contact is 3456789')
        print('son contact is 8765432')

parent = Parent()
child = Child()
for data in (parent,child):
    data.names()
    data.ages()
    data.contact()
```

```python
# polymorphism with inheritance

class Vehicle:
    def vehicletypes(self):
        print('many types of vehicles is there like cars,bikes,scotys,buses and lorrys')
    def wheels(self):
        print('different types of vehiclke having different wheeles example car having 4 bike having 2..')

class Car(Vehicle):
    def wheels(self):
        print('car having 4 wheels')

class Bike(Vehicle):
    def wheels(self):
        print('bike having 2 wheels')

class Lorry(Vehicle):
    def wheels(self):
        print('lorry having 12 wheels')

class Auto(Vehicle):
    def wheels(self):
        print('auto having 3 wheels')

object_Vehicle = Vehicle()
object_Car = Car()
object_Bike = Bike()
object_Lorry = Lorry()
object_Auto = Auto()

for w in (object_Vehicle,object_Car,object_Bike,object_Lorry,object_Auto):
    w.wheels()
```
