```python
# data Encapsulation and data Abstraction
'''
mainly using these concept is we can change our data into public and private
public data means we can access any one from inside the class and outside the class
    class ClasName:
        name = 'venkatesh' # public
privater data means its' confidemntial data that means we can acces inside the class members only
    class ClassName:
        __name = 'Venkatesh' # private
'''
class EmpData:
    name = 'venkatesh' # public data
    address = 'hyderabad'
    __salary = 289000
    __EmpId = '11y252dd' # private data
    
    def Profile(self):
        print('my name is:',self.name)
        print('my Employee Id is:',self.__EmpId)
        print('my Employee salary is:',self.__salary)

data = EmpData()
data.Profile()
print(data.name)# outside the class but accessing public data
print(data.address)# outside the class but accessing public data
print(data.__EmpId) # its a private data or confidentail data
print(data.__salary)
```

```python
class Data:
    def __init__(self,name,age,salary,address,contact,email,gender):
        self.name = name
        self.age = age
        self.__salary = salary
        self.address = address
        self.__contact = contact
        self.__email = email
        self.gender = gender
    def publicdata(self): # these function return publkic data
        print("my name is:",self.name)
        print('my address is:',self.address)
        print("my age is:",self.age)
        print("gender is:",self.gender)
    def privatedata(self): # these function return private data
        print('salary is:',self.__salary)
        print("my contact number is:",self.__contact)
        print("my email id is:",self.__email)
info = Data('divya',25,234567,'mumbai',9876543234,'divya@gmail.com','F')
info.privatedata() # these statement retunr calling the private data inside the class
print(info.name)
print(info.publicdata())
```

```python
# how can access the private data outside the class
class Information:
    name = 'chandra' # public data
    salary = 56780 # public data
    __contact = 9876543210 # private data
    __id = 'fghty1234f' # private data
    def accessData(self):
        print("my name is:",self.name)
        print("my salary is:",self.salary)
        print("my phone number is:",self.__contact)
        print("my Id is:",self.__id)
data = Information()
data.accessData()
```
