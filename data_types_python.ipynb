# Python Data Types

## Basic Data Types

Python supports several built-in data types including integers, floating-point numbers, strings, and booleans. Data types define the kind of values a variable can hold.

```python
# Integer example
a = 10
print(a, 'is of type', type(a))

# Float example
b = 3.75
print(b, 'is of type', type(b))

# String example
c = 'Python'
print(c, 'is of type', type(c))

# Boolean example
d = True
print(d, 'is of type', type(d))
```

## Collection Data Types

Collections group multiple values together. Common Python collection types are lists, tuples, dictionaries, and sets.

```python
# List example
fruits = ['apple', 'banana', 'cherry']
print('List:', fruits, 'type:', type(fruits))

# Tuple example
coordinates = (10, 20)
print('Tuple:', coordinates, 'type:', type(coordinates))

# Dictionary example
student = {'name': 'Ravi', 'age': 21}
print('Dictionary:', student, 'type:', type(student))

# Set example
colors = {'red', 'green', 'blue'}
print('Set:', colors, 'type:', type(colors))
```

## Problem Statement 1: Describe your profile

Create variables for your name, age, height, and whether you are a student. Print each value and its type.

## Type Conversion

Type conversion changes a value from one type to another. Use `int()`, `float()`, `str()`, and `bool()` to convert types.

```python
# Convert string to integer
price_text = '150'
price = int(price_text)
print(price, 'type:', type(price))

# Convert integer to string
quantity = 4
quantity_text = str(quantity)
print(quantity_text, 'type:', type(quantity_text))

# Convert number to boolean
is_available = bool(price)
print(is_available, 'type:', type(is_available))
```

## Mutable vs Immutable Types

Some data types can be changed after creation (mutable), such as lists and dictionaries. Immutable types, such as strings and tuples, cannot be changed once created.

```python
# Mutable list example
numbers = [1, 2, 3]
numbers.append(4)  # list can be modified
print('List after append:', numbers)

# Immutable tuple example
point = (5, 7)
print('Tuple:', point)
# point[0] = 6  # this would raise an error if uncommented
```
