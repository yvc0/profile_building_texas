## What is Python?

Python is a high-level, interpreted programming language known for readability, simplicity, and a large standard library. It supports multiple programming paradigms including procedural, object-oriented, and functional programming.

## Python Definitions

- **Interpreter**: A program that executes Python code line by line.
- **Syntax**: The set of rules that defines valid Python programs.
- **Variable**: A name that refers to a value stored in memory.
- **Expression**: A combination of values, variables, and operators that Python can evaluate to produce a result.

```python
# Example: simple Python statements
message = 'Hello, Python!'
number = 10
print(message)
print('The value of number is', number)
```

## Problem Statement 1: Print a welcome message

Write a Python script to print a welcome message and show the length of the message string.

```python
# Problem 1 solution
welcome_message = 'Welcome to Python programming!'
print(welcome_message)
print('Message length:', len(welcome_message))
```

## Code Comments

Comments are used to explain code. In Python, single-line comments start with `#`. Comments do not affect program execution.

```python
# This line defines a number variable
x = 5  # assign the integer 5 to x
# Print the value of x and the square of x
print('x =', x)
print('x squared =', x * x)
```

## Keywords and Identifiers

Python has reserved words called keywords that cannot be used as variable names. Identifiers are names used for variables, functions, and other objects.

```python
import keyword
print('Python keywords example:')
print(keyword.kwlist[:10])  # print first 10 keywords

# Valid identifiers
name = 'Alice'
age = 25
print('Name:', name, 'Age:', age)
```

## Problem Statement 2: Evaluate expressions

Write code that calculates the sum, difference, product, and quotient of two numbers, then prints each result with explanatory text.

```python
# Problem 2 solution
a = 12
b = 4
print('a + b =', a + b)
print('a - b =', a - b)
print('a * b =', a * b)
print('a / b =', a / b)
```
