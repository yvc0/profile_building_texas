### Conditional Statements In python or Control flow In python
1. if
2. elif
3. else

- if statement
    - if statement is used to decide whether a certain statement or block of statements will be executed or not 
- elif statements
    - more conditions we can perdform using elif statement
- else
    - condition is not satisfied then move to else statement

```python
# given name is valid or not
# python is followed indentations(program is written in perticular order)
'''syntax of if condition
if condition:
    statement1
    statement2
    statement3
    ....
'''
```

```python
# given number is even or not
a = 4
if (a%2==0): # 2%2
    print("given number is even")
    print("that number is ",a)
```

```python
# given number is even or odd number
a = 132445
if (a%2==0):
    print(a,"number is even")
else:
    print(a,"number is odd")
```

```python
# findout the student grades based on marks using conditional statements
# marks is (100-90)-A
# marks is (89-80)-B
# marks is (79-70)-C
# marks is (69-60)-D
# marks is (59-50)-E
# marks is below 50- Failed
s_mark = 50
if (s_mark<=100 and s_mark>=90): # that student got marks in between 100-90(True)
    print("student got A grade")
elif(s_mark<=89 and s_mark>=80):
    print("student got B grade")
elif(s_mark<=79 and s_mark>=70):
    print("student got C grade")
elif(s_mark<=69 and s_mark>=60):
    print("student got D grade")
elif(s_mark<=59 and s_mark>=50):
    print("student got E grade")
elif(s_mark<=49 and s_mark>=0):
    print("student got failed")
else:
    print("invalid marks")
```
