#### User defined functions
- these functions are created by user or ourself
    - example
        - ```
        def function_name(arguments or parameters):
            statement1
            statment2...
        function_name()
        ```
    - types
        1. with return statement
        2. without return statement
        3. required argument
        4. keyword arguments
        5. default arguments
        

````python
# addition of three numbers using function
def addition_three():
    a = int(input("enter first value"))
    b = int(input("enter second value"))
    c = int(input("enter third value"))
    return a+b+c
addition_three()
# return statmention that means function is ended
````

````python
# addition of three numbers using function
# we are passing outside varibles with in the function through arguments
# here a,b and c are arguments to functions
a = int(input("enter first value"))
b = int(input("enter second value"))
c = int(input("enter third value"))
def addition_three(a,b,c):
    return a+b+c
addition_three(a,b,c)
# return statmention that means function is ended
````

````python
# without return statement add three values result
def add_three():
    a = int(input("enter first value:"))
    b = int(input("enter second value:"))
    c = int(input("enter third value:"))
    print(a+b+c)
    print(a+c)
    print(a-b+c)
    print(a*b)
    print(a*b*c)
add_three()
    
````

````python
a = 12
b = 13
c = 14
def add_three(a,b,c):
    print("sum of a,b and c is ",a+b+c)
    return a+b+c
add_three(a,b,c)
````

````python
# keyword argument
name = input('enter username:')
age = int(input('enter user age:'))
email = input("enter user email:")
adr = input('enter user address')
def user_data(name,age,adr,email): # age=12
    print('my name is:',name)
    print('my age is :',adr)
    print('my address is:',age)
    print('my email id is:',email)
user_data('indu',12,adr,email)
````

````python
# keyword arguments type
def req(name,age,email,adr,contact): # name=venkat, age=venkat@gmail.com, email=28
    print('my name is:',name)
    print('my age is :',age)
    print('my address is:',adr)
    print('my email id is:',email)
    print("contact number is:",contact)
req(name='venkat',email='venkat@gmail.com',age=28,adr='hyderabad',contact='987654312')
# in this case we get error is required arguments error
````
