````python
# findout the evne numbers in between 1 to 1000
start_number = int(input("starting number:")) # 2
end_number = int(input("ending number:")) # 1000
while(start_number<=end_number): # 2<=1000
    if (start_number%2==0): # condition for even numbers
        print(start_number,end=" ")
    start_number+=1
````

````python
# findout the numbers in between 1 to 1000 where the numbers are divisible by 5 and not divisible 2
start_number = int(input("starting number:"))
end_number = int(input("ending number:"))
while (start_number<=end_number):
    if (start_number%5==0 and start_number%2!=0):
        print(start_number,end=",")
    start_number+=1 # start = start + 1
````

````python
# continue and break statements
# break - break statement stops the entire process of the loop
# continue - continue statements only stop the current iteration of the loop
````

````python
# break statement 
# i need to print 1 to 10 natural numbers
for i in range(1,10+1):
    if (i==7):
        break
    print(i,end=' ')
````

````python
# continue
for i in range(1,10+1):
    if (i==7):
        continue
    print(i,end=' ')
````

````python
# break statement in while loop
sn = int(input("starting number"))
en = int(input("ending number"))
while (sn<=en):
    if(sn==5):
        break
    print(sn)
    sn+=1
````
