```python
# random number- random means spmething that can not be predicted logically
import numpy as np
from numpy import random
```

```python
# i need to print any number randomly from 0 to 100
# random having lot of functionaliteis(rand,randint,choice...)
random_number = random.randint(100)
print(random_number)
```

```python
# i need to print any number randomly from 0 to 5
# random having lot of functionaliteis(rand,randint,choice...)
random_number = random.randint(5)
print(random_number)
```

```python
# rand function returns float values in between 0 to 1
rand_number = random.rand()
print(rand_number)
```

```python
# genrated 1-D array containg range mentioning in size method and range is 0 to 100
result = random.randint(100,size=(10))
print(result)
```

```python
# genrated 1-D array containg range mentioning in size method and range is 0 to 100
result = random.randint(10,size=(5))
print(result)
```

```python
# genrated 2-D array containg range mentioning in size method and range is 0 to 100
result = random.randint(100,size=(2,5))
print(result)
```

```python
# genrated 2-D array containg range mentioning in size method and range is 0 to 100
result = random.randint(100,size=(5,3))
print(result)
```

```python
# genrated 2-D array containg range mentioning in size method and range is 0 to 5
result = random.randint(5,size=(5,3))
print(result)
```

```python
# genrated 3-D array containg range mentioning in size method and range is 0 to 5
result = random.randint(10,size=(2,5,3)) # 2*5*3
# first 2 is 2-D arrays are 2
# in each 2D having again 1-D arrays are 5
# 3 is each 1-D having 3 elements
print(result)
```

```python
# in random number picking we have one option for select perticular elements from the array using choice
# choice - functionality used to pick number from list
choice_number = random.choice([2,4,6,8])
print(choice_number)
```

```python
# Create 2-D array using choice method
result = random.choice([2,4,6,8,10],size = (3,3))
print(result)
```

```python
ages = np.array([29,30,25,31,35,40,43])
condition = ages>30
result = ages[condition]
print(condition)
print(result)
```

```python
# getting even number ages from the array
ages = np.array([29,30,25,31,35,40,43])
condition = ages%2==0
result = ages[condition]
print(condition)
print(result)
```
