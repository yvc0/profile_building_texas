# Introduction to Machine Learning

Machine Learning is a branch of Artificial Intelligence where computers learn patterns from data and make predictions or decisions without being explicitly programmed for every task.

## Key Definitions

- **Dataset**: A collection of records used to train and test a model.
- **Feature**: An input variable such as age, income, or area.
- **Label**: The target value we want to predict such as price, spam, or category.
- **Model**: A mathematical structure that learns patterns from data.
- **Training**: Teaching the model using historical data.

## Supervised Learning

In supervised learning, the dataset contains input features and known output labels. The model learns from these examples and then predicts labels for new data.

## Example 1: Linear Regression for House Price Prediction

A real estate company wants to predict the price of a house based on its area. This is a regression problem because the output is a continuous number.

```python
import pandas as pd
from pathlib import Path
from sklearn.linear_model import LinearRegression

# Read the house price dataset
house_data = pd.read_csv(Path('house_price_hyderabad.csv'))
house_data.head()
```

```python
from sklearn.linear_model import LinearRegression

# Prepare features and target for house price prediction
X_house = house_data[['area_sqft', 'bedrooms', 'bathrooms', 'age_years', 'location_score']]
y_house = house_data['price_lakhs']

# Create and train the regression model
house_model = LinearRegression()
house_model.fit(X_house, y_house)

print('House price regression model trained successfully')
```

```python
import pandas as pd

# Dynamic testing for house price prediction
area_sqft = float(input('Enter area in sq ft: '))
bedrooms = int(input('Enter number of bedrooms: '))
bathrooms = int(input('Enter number of bathrooms: '))
age_years = int(input('Enter age of house in years: '))
location_score = float(input('Enter location score (1 to 10): '))

new_house = pd.DataFrame(
    [[area_sqft, bedrooms, bathrooms, age_years, location_score]],
    columns=['area_sqft', 'bedrooms', 'bathrooms', 'age_years', 'location_score']
)
predicted_price = house_model.predict(new_house)[0]
print('Predicted price for the entered house:', round(predicted_price, 2), 'lakhs')
```

## Example 2: Logistic Regression for Spam Detection

This is a classification problem because the output is a category: spam or not spam. Logistic regression is commonly used for binary classification.

```python
import pandas as pd
from pathlib import Path
from sklearn.linear_model import LogisticRegression

# Read the email spam dataset
email_data = pd.read_csv(Path('email_spam_data.csv'))
email_data.head()
```

```python
from sklearn.linear_model import LogisticRegression

# Prepare features and target for spam detection
X_email = email_data[['subject_length', 'sender_score', 'contains_offer', 'has_link']]
y_email = email_data['is_spam']

# Create and train the classifier
spam_classifier = LogisticRegression(max_iter=1000)
spam_classifier.fit(X_email, y_email)

print('Spam classifier trained successfully')
```

```python
import pandas as pd

# Dynamic testing for spam detection
subject_length = int(input('Enter subject length: '))
sender_score = float(input('Enter sender score (0 to 1): '))
contains_offer = int(input('Enter 1 if the email contains an offer, otherwise 0: '))
has_link = int(input('Enter 1 if the email contains a link, otherwise 0: '))

new_message = pd.DataFrame(
    [[subject_length, sender_score, contains_offer, has_link]],
    columns=['subject_length', 'sender_score', 'contains_offer', 'has_link']
)
prediction = spam_classifier.predict(new_message)[0]
print('Prediction for the entered message:', 'Spam' if prediction == 1 else 'Not Spam')
```

## Unsupervised Learning

In unsupervised learning, the data has no predefined labels. The model tries to find hidden groups or patterns in the data.

## Example 3: K-Means Clustering for Customer Segmentation

A company may want to group customers into segments based on age, income, and spending score. K-Means is a popular clustering algorithm used for this task.

```python
import pandas as pd
from pathlib import Path
from sklearn.cluster import KMeans

# Read the customer segmentation dataset
customer_data = pd.read_csv(Path('customer_segmentation_data.csv'))
customer_data.head()
```

```python
from sklearn.cluster import KMeans

# Prepare features for customer segmentation
X_customer = customer_data[['age', 'annual_income_k', 'spending_score', 'loyalty_points']]

# Create and train the clustering model
customer_kmeans = KMeans(n_clusters=2, random_state=42)
customer_kmeans.fit(X_customer)

print('Customer clustering model trained successfully')
```

```python
import pandas as pd

# Dynamic testing for customer segmentation
age = int(input('Enter customer age: '))
annual_income_k = float(input('Enter annual income in thousands: '))
spending_score = int(input('Enter spending score (0 to 100): '))
loyalty_points = int(input('Enter loyalty points: '))

new_customer = pd.DataFrame(
    [[age, annual_income_k, spending_score, loyalty_points]],
    columns=['age', 'annual_income_k', 'spending_score', 'loyalty_points']
)
cluster = customer_kmeans.predict(new_customer)[0]
print('Predicted cluster for the entered customer:', cluster)
```

## Important Difference Between Supervised and Unsupervised Learning

- Supervised learning uses labeled data.
- Unsupervised learning works on unlabeled data and finds hidden patterns.
