````python
# we can findout the higest salary and lowest salary with that person name
dict1 = {'venkat':30000,'joseph':45000,'aniket':50000,'sonam':100000}
list1 = []
for i in dict1.items():
    list1.append(i[1]) # i[0] having keys and i[1] having values
print(list1)
for j in dict1.items():
    if j[1]==max(list1): # who get maximum salary 100000==100000 False
        print(j[0],'getting maximum salary',j[1])
    elif j[1]==min(list1): # who get minimum salary 100000==30000 false
        print(j[0],'getting minimu salary',j[1])
        
        
````
