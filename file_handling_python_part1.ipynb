#### file handling in python
1. defination
    - collection of data or collection of information is nothing but file
    - examples - .py, .html, .py, .jpg, .png, .mp4, .mp3....
2. categories of files
    - create a file
    - write a data to file
    - read data from file
    - updated data to file
3. modes we can applying to files
    - create file using mode is 'x'
    - write a data to file using mode is 'w'
    - we need to update any existed file using mode is 'a'
    - read data from file using mode is 'r'
    - r+ read and write data into files
    - w+ write and read data into files
    - a+ append and read from the files.
4. synatx of files
    - fh = open('file name including location','mode')

```python
# create a file we need to apply x mode
fh = open('venkatesh_data.txt','x')
print('created a file successfully')
fh.close() # once the mode operation is completed then we need to close the file
```

```python
# write information to files - we can write data to existed file - w
fh1 = open('venkatesh_data.txt','w')
fh1.write('name:venkatesh \nemail:venkatesh@gmail.com \ncontact:+919494699851')# write method is add information to existed file
print("successfully write data to file")
fh1.close()
```

```python
# read data from existed file using r mode
fh2 = open('venkatesh_data.txt','r')
output = fh2.read()
print(output)
fh2.close()
```

```python
# read data from existed file using r mode
fh2 = open('C:/Users/yvenk/OneDrive/Documents/abc.txt','r')
#C:/Users/yvenk/OneDrive/Documents
output = fh2.read()
print(output)
fh2.close()
```

```python
# append data to existed files or updated information to existed files - a
fh3 = open('venkatesh_data.txt','a')
fh3.write('\naddress:hyderabad')
print('successfully updated data')
fh3.close()
```

```python
# read data
fh4 = open('venkatesh_data.txt','r')
result = fh4.read()
print(result)
fh4.close()
```

```python
# with key word we can perform file handling
# if you using file with keyword no need to mention close()
with open('indu.txt','x') as fh:
    print('successfully created filee')
    
```

```python
# add information to file
with open('indu.txt','w') as data:
    data.write('name:indu \nemail:indu@gmail.com \naddress:india')
    print("information successfully adding")
```
