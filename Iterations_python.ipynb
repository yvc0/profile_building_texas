#### Iterations or loop in python
- an iteration is an object that contains a countable number of values
- an iteration is an object that can be iterated upon meaning that you can traverse thruough all the values


```python
# syntax for iteration
# 1. for loop
# 2. while loop
'''for loop synatx
for iterata_variable in sequence_numbers:
    statements1
    statements2...
'''

'''while loop syntax

while (condition):
    statements1
    statements2..

'''
```

```python
# we need to print 1 to 100 natural numbers using for loop
# range takes starting ending-1 
# range(1,100) Starting 1 ending=100-1
for number in range(1,100+1): #in this condition starting is 1 end is 100
    print(number,end=',')
```

```python
# we are using conditional statements in loops
# print even numbers in between 500 to 700(700 is inclusive) using loops and control flow statements

for num in range(500,700+1):
    if (num%2==0):
        print(num,end=',')
```

```python
# print even number taking range values from user
start = int(input("starting value:"))
end = int(input("ending value:"))
for number in range(start,end+1):
    if (number%2==0):
        print(number,end=' ')
```

```python
# print numbers divisble by 2 & 3 and not divisble by 5 from 100 to 500 from user 
start = int(input("starting value:"))
end = int(input("ending value:"))
for i in range(start,end):
    if (i%2==0 and i%3==0):
        if (i%5!=0):
            print(i,end=' ')
```
