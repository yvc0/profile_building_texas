```python
class Profile:
    def __init__(self,name,salary,address,email): # parametrized constructor
        self.name = name
        self.salary = salary
        self.email = email
        self.address = address
    def myprofile(self):
        print('my name is:',self.name)
        print('my address is:',self.address)
    def mydata(self):
        print('my salary is:',self.salary)
        print('my email is:',self.email)
    def alldata(self):
        print('my name is:',self.name)
        print('my address is:',self.address)
        print('my salary is:',self.salary)
        print('my email is:',self.email)
# create object
data = Profile('divya',76278,'pune','divya@gmail.com') 
# createing constructor how many arguments are creating same values we are passing in with in the class
data.myprofile()
data.mydata()
data.alldata()
```

```python
# non parametrized constructor
class Employee:
    def __init__(self): # non Parametrized
        pass
    def employeedata(self,ename,eid,eemail,eaddress): # method
        print('employee Id:',eid)
        print('employee email:',eemail)
        print('employee Address:',eaddress)
        print('employee name:',ename)
# create object
emp = Employee()
emp.employeedata('venkat',1,'venkat@gmail.com','banglore')
```

```python
class Employees:
    def __init__(self,name,email,domine):
        self.name = name
        self.email = email
        self.domine = domine
    def employeinfo(self):
        print('employee name:',self.name)
        print('employee email:',self.email)
        print('employee domine:',self.domine)

# multiple objects
emp1 = Employees('venkatesh','venkatesh@gmail.com','python')
emp2 = Employees('joesph','joesph@gmail.com','Java')
emp3 = Employees('Aniket','Aniket@gmail.com','C++')
emp4 = Employees('Indu','indu@gmail.com','Data Engineer')
emp5 = Employees('Sonam','sonma@gmail.com','cloud')
emp6 = Employees('tanjeela','tanjeela@gmail.com','python')
emp7 = Employees('bhavya','bhavya@gmail.com','C')
print(emp7.employeinfo())
print(emp5.employeinfo())
print(emp3.employeinfo())
```
